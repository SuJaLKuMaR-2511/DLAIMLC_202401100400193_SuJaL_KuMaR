# **NLP - Text Preprocessing**

```python
nltk.download('punkt') # old version of nltk
nltk.download('punkt_tab') #latest version of nltk
```

```python
import nltk
from nltk.tokenize import word_tokenize, sent_tokenize;
```

# **1. Tokenization**

```python
text = "Natural Language Processing (NLP) is amazing! Let's explore it."
statement = "Please follow our 'blog' to see more information about new features, tips & tricks, and featured notebooks such as : /Analyzing a Bank Failure with Colab!."

# Word Tokenization
word_tokens = word_tokenize(statement)
print("Word Tokens:", word_tokens)

# Sentence Tokenization
sentence_tokens = sent_tokenize(statement)
print("Sentence Tokens:", sentence_tokens)
```

# **2a. LowerCasing in NLP**

```python
# Lowercasing each token
lower_tokens = [token.lower() for token in word_tokens]
print("Lowercased Tokens:", lower_tokens)
```

# **2b. UpperCasing in NLP**

```python
# Uppercasing each token
upper_tokens = [token.upper() for token in word_tokens]
print("Uppercased Tokens:", upper_tokens)
```

# **3. Stop Word Removal**

```python
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))
filtered_tokens = [token for token in word_tokens if token.lower() not in stop_words]
print("Filtered Tokens:", filtered_tokens)
```

# **4a. Stemming (Reducing Words to their Root Word)**

```python
from nltk.stem import PorterStemmer

ps = PorterStemmer()
stemmed_tokens = [ps.stem(token) for token in filtered_tokens]
print("Stemmed Tokens:", stemmed_tokens)
```

# **4b. Lemmatization (Reducing Words to their More Advanced Base Form)**

```python
from nltk.stem import WordNetLemmatizer

nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in filtered_tokens]
print("Lemmatized Tokens:", lemmatized_tokens)
```

# **5. Part-Of-Speech (POS) Tagging**

```python
nltk.download('omw-1.4')
```

```python
from nltk import pos_tag
nltk.download('averaged_perceptron_tagger')
# nltk.download('averaged_perceptron_tagger_eng')
```

```python
pos_tags = pos_tag(word_tokens)
print("POS Tags:", pos_tags)
```

# **6. Named Entity Recognition (NER)**

```python
from nltk.chunk import ne_chunk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
ner_tree = ne_chunk(pos_tags)
print("Named Entities:", ner_tree)
```

```python
import time
import sys
import os

def display_neon_author_animated(name):
    # Clear the terminal for a clean canvas
    os.system('cls' if os.name == 'nt' else 'clear')
    print("\n")

    # ANSI escape characters for styling
    cyan_bg = "\033[46m"
    black_fg = "\033[30m"
    bold = "\033[1m"
    reset = "\033[0m"
    shadow_char = "\033[90m░\033[0m"

    text = f"  AUTHOR: {name.upper()}  "
    length = len(text)

    # 1. Animate Top Padding (Slides in quickly)
    for _ in range(length):
        sys.stdout.write(cyan_bg + " " + reset)
        sys.stdout.flush()
        time.sleep(0.01)
    sys.stdout.write("\n")

    # 2. Animate Main Text Line (Types out character by character)
    for char in text:
        sys.stdout.write(cyan_bg + black_fg + bold + char + reset)
        sys.stdout.flush()
        time.sleep(0.04) # Slower delay for typing effect
    sys.stdout.write(shadow_char + "\n") # Add right-side shadow

    # 3. Animate Bottom Padding (Slides in quickly)
    for _ in range(length):
        sys.stdout.write(cyan_bg + " " + reset)
        sys.stdout.flush()
        time.sleep(0.01)
    sys.stdout.write(shadow_char + "\n") # Add right-side shadow

    # 4. Animate Bottom Shadow (Draws left to right)
    sys.stdout.write(" ") # Offset by 1 space to match the right-side drop shadow
    for _ in range(length):
        sys.stdout.write(shadow_char)
        sys.stdout.flush()
        time.sleep(0.015)
    sys.stdout.write("\n\n")

if __name__ == "__main__":
    display_neon_author_animated("SuJaL KuMaR")
```
