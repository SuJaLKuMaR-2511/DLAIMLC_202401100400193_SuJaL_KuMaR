# **1. Basic Word Embedding Example : Word -> Number/Vector**

```python
words = ["cat", "dog","apple"]

embeddings = {
    "cat": [1, 0, 0],
    "dog": [0, 1, 0],
    "apple": [0, 0, 1]
}

for word in words:
    print(embeddings[word])
```

# **2. Similarity**

```python
embeddings = {
    "cat": [0.9, 0.8, 0.1],
    "dog": [0.8, 0.9, 0.1],
    "apple": [0.1, 0.1, 0.9]
}

print("cat", embeddings["cat"])
print("dog", embeddings["dog"])
print("apple", embeddings["apple"])
```

# **3. Word2Vec**

```python
from gensim.models import Word2Vec

sentences = [
    ["cat", "eat", "milk"],
    ["dog", "eat", "food"],
    ["cat", "play", "ball"],
    ["dog", "play", "ball"]
]

model = Word2Vec(
    sentences,
    vector_size=2,
    window=2,
    min_count=1,
    sg=0
)

print(
    "Cat vector: ",
    model.wv["cat"]
)
print(
    "Dog vector: ",
    model.wv["dog"]
)
print(
    "Cat-Dog similarity: ",
    model.wv.similarity("cat", "dog")
)

print(
    "similar words: ",
    model.wv.most_similar("cat")
)
```

# **4. Glove**

```python
import gensim.downloader as api

glove = api.load("glove-wiki-gigaword-50")

print(
    glove["cat"],
    " ",
    glove.most_similar("cat")
)
```

# **5. FastText**

```python
from gensim.models import FastText

sentences = [
    ["I", "love", "machine", "learning"],
    ["I", "love", "deep", "learning"],
    ["Machine", "learning", "is", "fun"],
    ["Deep", "learning", "is", "fun"]
]

model = FastText(
    sentences,
    vector_size=2,
    window=2,
    min_count=1,
    sg=0
)

print(
    "Machine vector: ",
    model.wv["machine"]
)
print(
    "Learning vector: ",
    model.wv["learning"]
)
print(
    "Machine-Learning similarity: ",
    model.wv.similarity("machine", "learning")
)
```
